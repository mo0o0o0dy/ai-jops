```python
df.drop(['job_id', 'salary_currency', 'employee_residence', 'industry',
         'application_deadline', 'posting_date', 'job_description_length', 'benefits_score'], axis=1, inplace=True)
```

```python
df.head()
```

```python
def exp_level(any_value):
    if any_value == 'SE':
        return 'Senior'
    elif any_value == 'EN':
        return  'Entry'
    elif any_value == 'MI':
        return 'Mid-level'
    else:
        return 'Excutive'
```

```python
exp_level('SE')
```

```python
df.experience_level = df.experience_level.apply(exp_level)
```

```python
df.head()
```

```python
df.employment_type = df.employment_type.apply(lambda x:'Full Time' if x == 'FT' else 'Part Time' if x == 'PT' else 'Contract' if x == 'CT' else 'Freelance')
```

```python
df.head()
```

```python
df.head()
```

```python
def change_company_size(company_name):
    if company_name == "S":
        return "Small <50"
    elif company_name == "M":
        return "Medium 50-250"
    else:
        return "Large >250"
```

```python
df.company_size = df.company_size.apply(change_company_size)
```

```python
df.remote_ratio = df.remote_ratio.apply(lambda X : "No Remote" if X == 0 else "Hyrid" if X == 50 else "Remote")
```
